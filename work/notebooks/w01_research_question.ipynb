````python
# This cell is for CODE (numbers, a query, a check).
# Write your text answer in the cell ABOVE this one — typing sentences here breaks Run All.

## 2. The question: decision, action, cost of a wrong call

#**Search question:** Given a page's current performance and content signals, how urgently does it need a content refresh, and why?

#**Unit of analysis:** one page (`content_id`), scored at a point in time.

#**Output:** an urgency/opportunity score per page (0-100), plus a reason code (e.g. `stale_visible_page`, `declining_with_demand`, `low_ctr_visible_page`) explaining *why* it scored the way it did — not just a number, but something a reviewer can inspect and trust.

#---

#**What decision does this improve?**

#Right now, deciding which pages to refresh out of a large inventory is done by manual spot-checks — someone notices a page slipping, or checks a handful they remember. That doesn't scale past a few dozen pages, and it has no consistent standard for "worth reviewing." This work turns that into a repeatable weekly triage: a ranked queue that says, out of everything, review these N pages first.

#**Who acts on it?**

#A content strategist or editor with a fixed weekly review capacity — say, time to properly review 20-50 pages a week out of a much larger inventory. They pull the top of the ranked queue into that week's refresh work instead of picking pages by memory or gut feel.

#**What does a wrong recommendation cost?**

#The cost isn't symmetric, and I think that matters for how I'll eventually tune the score:

#- **False positive** (the model flags a page as urgent, but it wasn't really): the editor spends an hour or two refreshing a page that didn't need it. That's a real cost — wasted editor time is expensive — but it's a *soft*, recoverable cost. The page still exists, nothing broke, the team just spent effort somewhere with lower payoff than ideal.
#- **False negative** (a page is genuinely declining, but the model doesn't flag it): the page keeps losing visibility and traffic silently, week after week, until someone happens to notice it manually — which might be a long time, since that's exactly the failure mode this project exists to fix. This is a *slower, compounding* cost: the longer it goes unnoticed, the more cumulative traffic and ranking position gets lost.

#Because a missed decline compounds silently while a wasted review is a one-time, bounded loss, I'd lean toward tuning the scoring to accept somewhat more false positives in exchange for fewer false negatives — recall matters more here than raw precision, within the limits of what the editor's actual weekly capacity can absorb (there's no point flagging 500 "urgent" pages if only 50 will ever get reviewed).

#**Why this needs to be an explicit tradeoff, not an afterthought:** the lane guide's Section 11 makes the same point generally — every threshold I pick (how big a drop counts as decline, how low a CTR counts as weak, how many pages fit review capacity) is a policy choice I'm making, not a fact the data hands me. I'll need to write down which tradeoff I accepted and why, not just report a single "optimal" score.
````

````python
# This cell is for CODE (numbers, a query, a check).
# Write your text answer in the cell ABOVE this one — typing sentences here breaks Run All.

## 4. Careful words: what I can and can't claim

#**What my work will be able to say:**

#- **Observed:** "54.2% of pages in this dataset are currently trending down" and "32.5% of visible pages have CTR under 0.5%" — these are direct measurements from the data, not inferences.
#- **Directional:** "Pages with this combination of signals (declining trend + real demand + weak CTR) are *more likely* to be worth reviewing than pages without it." This is a pattern the model finds in the data, not a certainty about any individual page.
#- **Decision-support:** "This ranked queue orders candidates so a reviewer spends limited time on the most promising pages first." The score supports a human decision — it doesn't replace the reviewer's judgment, and it doesn't act on its own.
#- A score reflects "the right page to review first," not "a page guaranteed to recover if someone edits it" — that distinction (Section 6 of the lane guide) is one I'll repeat in my final write-up, not just note here.

#**What my work will never claim:**

#- **Causal proof.** I can't say a refresh *caused* a recovery. Proving that would require an actual experiment — refreshing some pages and holding others back as a control — which is outside what this data alone can support. If a page I flag as urgent later recovers after being refreshed, that's a correlation I can report, not evidence I made it happen.
#- **"Predicting Google."** I'm not modeling or claiming insight into Google's actual ranking algorithm. My signals (impressions, clicks, position, CTR, trend) are *observed outcomes* of whatever Google's algorithm already did — not a window into how it works.
#- **That a decline is definitely a decline.** Section 7 of the lane guide lists several look-alikes I have to actively rule out before trusting a "declining" label: consolidation (a sibling page absorbed the traffic), seasonality (the topic naturally dips on a calendar), SERP/AI click loss (impressions and position held steady but clicks dropped because the search results page itself changed), and plain noise (low-volume wiggle with no real pattern). A page flagged as "declining" without checking these could just be wrong.
#- **That results from the 30,000-row starter slice generalize to the full ~79M-row warehouse** without being re-validated there. The starter numbers are real, but they're a small anonymized sample, tested with client-holdout validation on that sample only.
#- **Anything that could re-identify a real client, domain, URL, or query.** Every output stays at the level of pseudonymized IDs and aggregated metrics — per the public-safe rules in Section 14, since the eventual deliverable is a public research paper.

#The throughline: this project produces evidence-backed priorities for a human reviewer, not verdicts. Every claim gets hedged to match what the data can actually support — observed, directional, decision-support — and nothing stronger than that.#
````
